# Improving Conversion Rate through ML-Driven Lead Scoring for Term Deposits

**Understanding The Business in General**

Term deposit (deposito berjangka) adalah salah satu sumber dana (likuiditas) yang sangat krusial bagi bank institusi ritel. Bank membutuhkan dana ini untuk kemudian disalurkan kembali dalam bentuk pinjaman atau kredit.

Untuk menjaring nasabah agar mau membuka deposito, bank ini menggunakan metode Direct Telemarketing. Tim sales atau call center akan secara proaktif menelepon daftar nasabah yang ada untuk menawarkan produk tersebut.

Telemarketing sangat bergantung pada interaksi manusia. Keberhasilan kampanye (nasabah membuka Deposit Berjangka sebagai result "Yes") sangat bergantung pada profil nasabah, kondisi ekonomi saat itu, dan efektivitas agen yang menelepon.

**Problem in Business**

Kami menyoroti bahwa tingkat konversi yang membuka Term Deposit dari campaign telemarketing sekarang sebesar 11.3%. Rata-rata nilai konversi telemarketing (leads based-call) adalah 10-15%, yang mana menunjukkan bahwa tingkat konversi nasabah yang akhirnya membuka Term Deposit sejak campaign telemarketing sudah berada di nilai rata-rata.

(referensi: https://bytewrites.org/telemarketing-b2b-lead-generation/)

Akan tetapi, dari segi bisnis telemarketing masih menghabiskan 88.7% waktu mereka untuk berbicara pada nasabah yang pada akhirnya tidak membuka Term Deposit.

**Goals**

1. Propensity Model (Prediksi Kecenderungan Nasabah Membuka Term Deposit). Membangun model machine learning untuk memprediksi probabilitas nasabah membuka Term Deposit, sehingga bank dapat mengidentifikasi dan memprioritaskan nasabah dengan peluang konversi tertinggi sebagai target utama campaign telemarketing.

2. Meningkatkan efektivitas dan efisiensi campaign telemarketing dengan memfokuskan upaya penawaran kepada top-tier leads yang memiliki probabilitas tinggi untuk membuka Term Deposit. Pendekatan ini diharapkan dapat meningkatkan conversion rate sekaligus mengurangi waktu dan biaya yang dihabiskan untuk menghubungi nasabah dengan peluang konversi rendah.

3. Mengembangkan model klasifikasi prediktif (ML-Driven Lead Scoring) yang menganalisis karakteristik serta riwayat interaksi nasabah untuk menghasilkan skor probabilitas pembukaan Term Deposit. Skor tersebut digunakan sebagai dasar dalam menentukan prioritas prospek yang akan dihubungi oleh tim telemarketing.

4. Mengimplementasikan sistem lead prioritization yang menghasilkan daftar top-tier leads bagi tim telemarketing. Dengan memprioritaskan nasabah yang memiliki peluang konversi tinggi dan mengurangi panggilan kepada prospek dengan peluang rendah.

**Matrics Evaluation**

Type 1 eror: False Positive (FP)
* Nasabah tidak ingin melakukan deposit tetapi campaign tetap dikirimkan
* Konsekuensi: Sia-sianya waktu dan resource yang digunakan untuk menawarkan campaign ini.

Type 2 eror: False Negative (FN)
* Nasabah ingin melakukan deposit tetapi tidak ditargetkan dalam campaign
* Konsekuensi: Kehilangan calon potensial nasabah yang akan melakukan deposit.

F0.5-score dipilih karena memberikan bobot yang lebih besar pada Precision dibanding Recall artinya, model lebih diutamakan untuk memastikan bahwa nasabah yang diprediksi akan membuka deposito memang benar-benar memiliki peluang tinggi untuk melakukan deposit.

Hal ini didasari pada pernyataan sebelumnya yaitu Term Deposit dari campaign telemarketing hanya sebesar 11.3% dan menghabiskan 88.7% cost untuk berbicara pada nasabah yang pada akhirnya tidak membuka Term Deposit. DiKarenakan setiap panggilan telemarketing membutuhkan waktu, biaya operasional, dan tenaga sales, maka False Positive menjadi lebih mahal.

**Benefit**

Salah satu dari benefit utama dengan perbaikan nilai konversi yang didapat adalah menurunnya Cost Per Acquisition (CPA) tanpa menurunkan jumlah penyedia liquiditas yang stabil secara signifikan.    

For starter, nasabah-nasabah yang membuka Term Deposit akan mengunci dana nya di deposit pada jenjang waktu tertentu, yang tentunya akan menjadikan diri mereka sebagai penyedia liquiditas yang stabil.    

Lalu, dari segi CPA pun bila kita menggunakan nilai Customer Acquisition Cost (CAC) berdasarkan referensi berikut (https://ijefm.co.in/v8i8/Doc/24.pdf) di mana cost per nasabah adalah €130.   
Bila diasumsikan dengan current tingkat konversi sebesar 11.3%, maka:   
- Bila 100 orang ditelpon, hanya akan ada 11 orang yang pada akhirnya membuka term deposit. Tetapi kita sudah menghabiskan cost sebesar €13.000 (€130 * 100 orang).
Dengan memperbaiki tingkat konversi kita, andaikan kita berhasil memperbaiki tingkat konversi menjadi 30%, maka:    
- Bila ada 30 orang ditelpon setelah disaring, dan didapatkan 11 orang yang sama membuka term deposit, maka kita dapat memangkas CAC nya lebih dari 60% ((€3.900-€13.000)/€13.000).

Kondisi data yang tidak seimbang ini memiliki beberapa implikasi penting untuk pemodelan *machine learning*:

1.  **Bias Model**: Model _machine learning_ cenderung akan bias terhadap kelas mayoritas ('no') karena data latihannya didominasi oleh kelas tersebut. Akibatnya, model mungkin akan memprediksi sebagian besar kasus sebagai 'no' untuk mencapai akurasi keseluruhan yang tinggi, tetapi gagal dalam mengidentifikasi kelas minoritas ('yes') yang seringkali menjadi kelas target yang lebih penting bagi bisnis.
2.  **Performa Metrik Evaluasi**: Metrik evaluasi standar seperti accuracy bisa menyesatkan. Model yang memprediksi semua sebagai 'no' akan memiliki akurasi ~88.7%, padahal model tersebut sama sekali tidak berhasil mengidentifikasi nasabah yang melakukan deposit. Metriks seperti **Precision, Recall, F Beta Score**  menjadi jauh lebih relevan untuk mengevaluasi performa model pada data yang tidak seimbang, karena model yang baik harus mampu mengidentifikasi kelas minoritas dengan efektif.

### Pdays Analisis

```python
df_bank[df_bank['pdays'] == -1].groupby(['previous', 'poutcome']).size().reset_index(name='count (-1)')
```

* **previous = 0 dan poutcome = 'nonexistent':** Terdapat **35,551 nasabah**. Ini adalah nasabah yang belum pernah dihubungi sebelumnya (atau belum ada riwayat kontak yang tercatat) dan belum ada hasil kampanye sebelumnya.
* **previous > 0 dan poutcome = 'failure':** **Terdapat 4,110 nasabah** (jumlah dari baris previous 1 sampai 6 dengan poutcome 'failure'). Ini adalah nasabah yang pernah dihubungi sebelumnya, tetapi kampanye tersebut gagal, dan mereka tidak dihubungi kembali pada kampanye terakhir.


```python
# pdays == -1 & previous == 0 -> Benar-benar baru, belum pernah dikontak sama sekali
# pdays == -1 & previous > 0  -> Pernah dikontak di masa lalu, namun tidak di kampanye terakhir
conditions = [
    (df_bank['pdays'] == -1) & (df_bank['previous'] == 0),
    (df_bank['pdays'] == -1) & (df_bank['previous'] > 0)
]
choices = ['new prospects', 'unreachable']

# default='Contacted' artinya jika pdays != -1 (baru saja dikontak di kampanye saat ini)
pdays_segment_values = np.select(conditions, choices, default='contacted')

# Tampilan jumlah dan  persentase dari ringkasan customer segment
segment_counts = pd.Series(pdays_segment_values).value_counts()
segment_percentages = pd.Series(pdays_segment_values).value_counts(normalize=True) * 100

segment_pdays = pd.DataFrame({
    'Count': segment_counts,
    'Percentage': segment_percentages
})

segment_pdays
```

* **New Prospects (35,551 nasabah, 86.34%):** Ini adalah nasabah yang pdaysnya -1 dan previousnya 0. Artinya, mereka belum pernah dihubungi sama sekali pada kampanye sebelumnya.Jadi  mereka adalah "prospek baru" yang belum memiliki riwayat kontak.
* **Unreachable (4,110 nasabah, 9.98%):** Nasabah dalam kategori ini memiliki pdays -1 dan previous > 0. Ini berarti mereka pernah dihubungi pada kampanye-kampanye sebelumnya, namun tidak pada kampanye terakhir. Maka mereka dianggap "tidak dapat dijangkau" atau tidak merespons pada upaya kontak kampanye - kampanye sebelumnya.
* **Contacted (1,515 nasabah, 3.68%):** Kelompok ini adalah nasabah yang pdaysnya bukan -1 (artinya baru saja dihubungi pada kampanye saat ini). Ini adalah segmen yang menunjukkan aktivitas kontak paling baru.

Distribusi Pelanggan berdasarkan pdays_bin dan Hasil Deposit:

* **Nasabah yang baru dihubungi kembali (pdays_bin 0-7, 8–14, 15–21, 22–28)**: Kelompok ini menunjukkan tingkat konversi deposit ('yes') yang sangat tinggi. Misalnya, untuk pdays_bin 0-7, sekitar 65.76% nasabah melakukan deposit. Ini mengindikasikan bahwa nasabah yang baru saja dihubungi kembali (dalam rentang 28 hari) memiliki kemungkinan lebih besar untuk melakukan deposit.
* **Nasabah yang belum pernah dihubungi (pdays_bin never_contacted)**: Kelompok ini merupakan mayoritas dataset. Namun, proporsi deposit 'yes' pada kelompok ini sangat rendah, hanya sekitar 9.26%. Ini menegaskan bahwa nasabah yang belum pernah dihubungi sebelumnya atau sudah lama tidak dihubungi memiliki kemungkinan yang sangat kecil untuk melakukan deposit.

Jarak waktu sejak kontak terakhir (pdays_bin) merupakan indikator kuat keberhasilan deposit. Nasabah yang baru saja dihubungi kembali memiliki peluang konversi yang jauh lebih tinggi. Fokus kampanye telemarketing pada nasabah yang memiliki riwayat kontak baru-baru ini dapat meningkatkan efisiensi secara signifikan.

```python
# Buat tabel kontingensi
contingency_table = pd.crosstab(df_bank['pdays_bin'], df_bank['deposit_num'])

# Lakukan uji Chi-Square
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Test antara Pdays dan Hasil Deposit:")
print(f"Statistik Chi-Square: {chi2:.4f}")
print(f"P Value: {p_value}")

# Interpretasi hasil
alpha = 0.05
if p_value < alpha:
    print(f"\nKarena nilai P ({p_value}) kurang dari tingkat signifikansi ({alpha}), maka")
    print("tolak Hipotesis Nol: Ada hubungan yang signifikan secara statistik antara 'pdays' dan 'deposit'.")
elif p_value >= alpha:
    print(f"\nKarena nilai P ({p_value}) lebih besar dari atau sama dengan tingkat signifikansi ({alpha}), maka")
    print("terima Hipotesis Nol: Tidak ada hubungan yang signifikan antara 'pdays' dan 'deposit'.")
```

```python
# 1. Pipeline untuk nominal (One-Hot Encoding)
onehot_pipe = ImbPipeline([
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', drop='if_binary', sparse_output=False))
])

# 2. Pipeline untuk multikelas dengan banyak kategori unik (Binary Encoding)
binary_pipe = ImbPipeline([
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('binary', BinaryEncoder())
])

# 3. Pipeline untuk fitur numerik (Median Imputation & Scaling)
num_pipe = ImbPipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# 4. Menggabungkan seluruh pipa ke dalam satu Transformer besar
preprocess = ColumnTransformer([
    ('onehot', onehot_pipe, nominal_cols),
    ('binary', binary_pipe, multiclass_cols),
    ('num', num_pipe, numeric_cols)
])
```

Dalam proyek _machine learning_ ini, kami menggunakan **pipeline pre-processing** untuk memastikan data siap digunakan oleh model dan untuk menghindari _data leakage_. Pipeline ini dibangun menggunakan `ColumnTransformer` yang memungkinkan penerapan transformasi yang berbeda pada kolom-kolom yang berbeda dalam dataset secara paralel.

Berikut adalah komponen-komponen utama dari pipeline pra-pemrosesan `preprocess`:

1.  **`onehot_pipe` (untuk Fitur Kategorikal Nominal)**: Bertujuan untuk mengubah fitur kategorikal nominal (tanpa urutan) menjadi representasi numerik menggunakan One-Hot Encoding. Ini mencegah model mengasumsikan hubungan ordinal yang tidak ada antar kategori.
    *   **Fitur yang Diterapkan:** `marital`, `education`, `credit`, `housing`, `loan`, `contact`, `month`, `day_of_week`, `poutcome`.


2.  **`binary_pipe` (untuk Fitur Kategorikal dengan Kardinalitas Tinggi)**: Bertujuan untuk mengubah fitur kategorikal dengan banyak kategori unik (kardinalitas tinggi) menjadi representasi biner. Ini dapat mengurangi dimensi fitur dibandingkan One-Hot Encoding pada fitur dengan banyak kategori, serta mempertahankan beberapa informasi struktural.
    * **Fitur yang Diterapkan:** `job`.

3.  **`num_pipe` (untuk Fitur Numerik Kontinu)**: Bertujuan untuk mengisi nilai yang hilang dan menstandarisasi fitur numerik.
    *   **Fitur yang Diterapkan:** `age`, `campaign`, `pdays`, `previous`, `emp_var_rate`, `cons_price_idx`, `cons_conf_idx`, `euribor3m`, `nr_employed`.

4.  **`preprocess` (ColumnTransformer)** adalah kolom pengatur yang menggabungkan semua pipeline di atas. Ini memungkinkan setiap set kolom (nominal, multiclass, numerik) untuk diproses secara independen dan hasilnya kemudian digabungkan menjadi satu matriks fitur yang telah diproses. Ini sangat efisien karena menghindari kebutuhan untuk memanipulasi dataframe secara manual untuk setiap jenis fitur.

Teknik re-sampling untuk mengatasi imbalance data projek ini menggunakan:
1. SMOTE (Synthetic Minority Over-sampling Technique): Membuat sampel baru buatan (synthetic) di sepanjang garis yang menghubungkan sampel-sampel minoritas terdekat.
2. Tomek Links: Menghapus sampel kelas mayoritas yang posisinya sangat dekat dengan sampel kelas minoritas untuk mempertegas batas keputusan (decision boundary).
3. Edited Nearest Neighbors (ENN): Menghapus sampel yang kelasnya berbeda dari mayoritas tetangga terdekatnya untuk membersihkan noise pada data.

Dari hasil objektif tersebut maka dipilihlah **LightGBM dengan TomekLinks** untuk dilanjutkan ke tahap threshold tuning dan data test.

##  Threshold Tuning

```python
# 1. Ambil model terbaik hasil tuning Anda
best_pipeline = random_search_tomek.best_estimator_

# 2. Dapatkan probabilitas Out-of-Fold menggunakan cross_val_predict
print("Menghitung probabilitas Out-of-Fold untuk data Training...")
y_train_proba = cross_val_predict(
    best_pipeline,
    X_train,
    y_train,
    cv=cv_stratified,
    method="predict_proba",
    n_jobs=-1
)[:, 1]

# 3. Hitung precision, recall, dan threshold untuk kurva
precisions, recalls, thresholds = precision_recall_curve(y_train, y_train_proba)

# 4. Rumus diubah khusus untuk F0.5-Score (Beta = 0.5)
# F0.5 = (1 + 0.25) * P * R / (0.25 * P + R)
f05_scores = (1.25 * precisions * recalls) / ((0.25 * precisions) + recalls + 1e-10)

# Cari index dengan F0.5-score tertinggi
best_threshold_idx = np.argmax(f05_scores)
best_threshold = thresholds[best_threshold_idx]
best_f05_curve = f05_scores[best_threshold_idx]

# 5. Visualisasi Kurva F0.5
plt.figure(figsize=(10, 5))
plt.plot(thresholds, f05_scores[:-1], "g-", label="F0.5 Score", linewidth=2) # Warna hijau
plt.axvline(x=best_threshold, color="r", linestyle="--", label=f"Best Threshold = {best_threshold:.4f}")
plt.xlabel("Threshold")
plt.ylabel("F0.5 Score")
plt.title("F0.5 Score vs Decision Threshold (Cross-Validated)")
plt.legend(loc="lower right")
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(np.arange(0, 1.1, 0.1)) # Menampilkan threshold dari 0.0 hingga 1.0 dengan interval 0.1
plt.show()

print(f"Threshold Terbaik berdasarkan PR-Curve: {best_threshold:.4f} dengan F0.5: {best_f05_curve:.4f}")
print(f"Parameter Terbaik dari Tuning Tomek: {random_search_tomek.best_params_}")
```

Hasil dari *threshold tuning* menunjukkan bahwa **_threshold_ terbaik berdasarkan kurva _Precision-Recall_ adalah 0.4925, dengan nilai _F0.5-Score_ sebesar 0.5178.**

1.  **Memaksimalkan _F0.5-Score_**: Proses _threshold tuning_ dilakukan dengan mengevaluasi _F0.5-Score_ pada berbagai nilai _threshold_ (probabilitas). *Threshold* 0.4925 adalah titik di mana model mencapai keseimbangan terbaik antara _Precision_ dan _Recall_ yang diinginkan, sehingga menghasilkan _F0.5-Score_ tertinggi.

2.  **Pengurangan _False Positives_**: Dengan memilih *threshold* ini, model diharapkan dapat lebih selektif dalam mengklasifikasikan nasabah sebagai 'akan deposit' (_positive_). Ini akan membantu meminimalkan jumlah _False Positives_ (nasabah yang diprediksi akan deposit tetapi sebenarnya tidak), yang pada akhirnya akan mengurangi pemborosan waktu dan biaya bagi tim telemarketing.

```python
tn, fp, fn, tp = cm.ravel()

# Total database nasabah yang disimulasikan, menggunakan total dari data test
total_database = tn + fp + fn + tp

# Biaya operasional per nasabah yang dihubungi berdasarkan referensi jurnal (€130)
cost_per_contact = 130

# Performa Model dari hasil Data Test Anda
baseline_conversion = 0.113  # 11.3% tingkat konversi awal
precision_test = tp / (tp + fp)    # Precision model LightGBM setelah tuning dari data test
recall_test = tp / (tp + fn)       # Recall model LightGBM setelah tuning dari data test


# SKENARIO LAMA (TANPA ML)
nasabah_dihubungi_lama = total_database
sukses_likuiditas_lama = int(total_database * baseline_conversion)
total_cost_lama = nasabah_dihubungi_lama * cost_per_contact
cac_lama = total_cost_lama / sukses_likuiditas_lama

# SKENARIO BARU (DENGAN ML LIGHTGBM)
# Jumlah sukses didapat langsung dari True Positives (TP)
sukses_likuiditas_baru = tp

# Jumlah yang perlu ditelepon dihitung dari TP + FP (yang diprediksi 'yes')
nasabah_dihubungi_baru = tp + fp
total_cost_baru = nasabah_dihubungi_baru * cost_per_contact
cac_baru = total_cost_baru / sukses_likuiditas_baru

# METRIK EFISIENSI
penghematan_biaya = total_cost_lama - total_cost_baru
persentase_hemat_biaya = (penghematan_biaya / total_cost_lama) * 100
penurunan_cac = ((cac_baru - cac_lama) / cac_lama) * 100
efisiensi_waktu = ((nasabah_dihubungi_lama - nasabah_dihubungi_baru) / nasabah_dihubungi_lama) * 100

# MEMBUAT TABEL RINGKASAN BREADOWN
cba_data = {
    "Komponen Analisis": [
        "Total Basis Data (Nasabah)",
        "Jumlah Nasabah Dihubungi",
        "Tingkat Konversi / Presisi",
        "Total Likuiditas Sukses Diraih (Nasabah)",
        "Total Biaya Pemasaran (€)",
        "Customer Acquisition Cost / CAC (€)"
    ],
    "Strategi Lama (Tanpa ML)": [
        total_database,
        nasabah_dihubungi_lama,
        f"{baseline_conversion*100:.1f}%",
        sukses_likuiditas_lama,
        f"€{total_cost_lama:,.0f}",
        f"€{cac_lama:,.2f}"
    ],
    "Strategi Baru (Dengan ML LightGBM)": [
        total_database,
        nasabah_dihubungi_baru,
        f"{precision_test*100:.1f}%",
        sukses_likuiditas_baru,
        f"€{total_cost_baru:,.0f}",
        f"€{cac_baru:,.2f}"
    ]
}

df_cba = pd.DataFrame(cba_data).set_index("Komponen Analisis")

# Menampilkan Tabel Berwarna untuk Jupyter / Colab
print("TABEL SIMULASI COST-BENEFIT ANALYSIS (CBA)")
display(df_cba)

print("\n" + "="*75)
print("                       KESIMPULAN MATRIKS UTAMA                       ")
print("="*75)
print(f"-> Total Penghematan Anggaran Pemasaran : €{penghematan_biaya:,.0f} ({persentase_hemat_biaya:.1f}% Lebih Hemat)")
print(f"-> Pemangkasan Waktu Telemarketing      : {efisiensi_waktu:.1f}% Waktu Agen Diselamatkan")
print(f"-> Penurunan Biaya Akuisisi (CAC Unit)  : {abs(penurunan_cac):.1f}% Lebih Murah Per Nasabah")
print("="*75)
```
